## Ejercicio 1

```python
def dic_cad_txt(cadenaTxt):
    """Esta función cuenta las diferentes letras que hay en la cadena de texto que recibe y devuelve un diccionario con la letra como clave y el número de veces que aparece esta como valor

    Args:
        cadenaTxt (str): Cadena de texto para contabilizar las diferentes letras que tiene y cuantas veces aparecen cada una

    Returns:
        dict: Devuelve un diccionario con clave la letra y como valor el número de veces que aparece en la cadena de texto que se ha recibido
    """
    dict_final = {}
    for letra in cadenaTxt:
        if letra != " ":
            if letra in dict_final:
                dict_final[letra] += 1
            else:
                dict_final[letra] = 1
    return dict_final

print(dic_cad_txt("Hola mi nombre es Jairo Martín Lucas"))

```

```python
nums = [1, 2, 3, 6, 8, -10, 9, -41, 33]
nums2 = [5, 28, 19, 5, 33, -20, 12, -3, 11]
def calc_dif(num, num2):
    """Calcula la diferencia entre un número y otro (da igual el orden, el resultado siempre será positivo)

    Args:
        num (int): Primer número para calcular la diferencia con el segundo
        num2 (int): Segundo número para calcular la diferencia con el primero

    Returns:
        int: Diferencia entre los números recibidos
    """
    if num < num2:
        return num2 - num
    else:
        return num - num2
diferencia = list(map(calc_dif, nums, nums2))
print(diferencia)
```

## Ejercicio 5

```python
notas = [1, 7, 3, 6, 8, 0, 9, 4, 0]
def calc_aprob(notas, nota_aprobado = 5):
    """Con la lista de notas que recibe calcula el promedio y dependiendo de la nota de aprobación recibida debuelve 'Aprobado' o 'Suspenso' y la nota media

    Args:
        notas (list[int]): Lista de notas para calcular el promedio
        nota_aprobado (int, optional): Nota promedio de aprobado. Defaults to 5.

    Returns:
        tuple: Devuelve un 'Aprobado' o 'Suspenso' y la nota media
    """
    suma = 0
    for nota in notas:
        suma += nota
    media = suma / len(notas)
    if media >= nota_aprobado:
        return ("Aprobado", media)
    else:
        return ("Suspenso", media)
print(calc_aprob(notas, 4))
calc_aprob(notas)
```

## Ejercicio 32

```python
empleados = [
    {"nombre": "Ana García", "puesto": "Gerente"},
    {"nombre": "Luis Martínez", "puesto": "Desarrollador"},
    {"nombre": "Ana López", "puesto": "Diseñadora"},
    {"nombre": "Carlos García", "puesto": "Analista"},
    {"nombre": "Sofía Hernández", "puesto": "Recursos Humanos"}
]

def encontrar_trabajador(empleados):
    buscar_trabajador = input("Introduzca el nombre o apellido del trabajador que quiere buscar: ").lower()
    resultados = []
    for trabajador in empleados:
        if buscar_trabajador in trabajador.get("nombre").lower():
            resultados.append(trabajador.get("nombre") + " trabaja de " + trabajador.get("puesto"))

    if len(resultados) >= 1:
        return "\n".join(resultados)
    else:
        return buscar_trabajador + " no trabaja aquí"

print(encontrar_trabajador(empleados))
```

## Ejercicio 34

```python
class Arbol:
    def __init__(self, tronco, ramas):
        self.tronco = tronco
        self.ramas = ramas
    def crecer_tronco(self, longitud_creciada):
        self.tronco += longitud_creciada
    def nueva_rama(self, longitud_rama):
        self.ramas.append(longitud_rama)
    def crecer_ramas(self, longitud_crecida):
        for rama in self.ramas:
            rama += longitud_crecida
    def quitar_rama(self, posicion_rama):
        self.rama.pop(posicion_rama)
    def info_arbol(self):
        print("Longitud tronco: " + str(self.tronco))
        cont = 1
        for rama in self.ramas:
            print("Rama: " + str(cont) + ", longitud: " + str(rama))
            cont += 1
arbol1 = Arbol(1, [])
arbol1.crecer_ramas(1)
arbol1.nueva_rama(1)
arbol1.nueva_rama(2)
arbol1.crecer_ramas(1)
arbol1.info_arbol()
```

## Ejercicio 35

```python
class UsuarioBanco:
    def __init__(self, nombre, saldo, cuenta):
        self.nombre = nombre
        self.saldo = saldo
        self.cuenta = cuenta
    def retirar_dinero(self, dinero_sacar):
        if self.saldo > dinero_sacar:
            self.saldo -= dinero_sacar
            return "Dinero sacado: " +  str(dinero_sacar) + ". Saldo: " + str(self.saldo)
        else:
            return "Saldo insuficiente. " + "Saldo: " + str(self.saldo)
    def transferir_dinero(self, dinero_transferir, persona):
        if self.saldo > dinero_transferir:
            self.saldo -= dinero_transferir
            persona.saldo += dinero_transferir
            return "Dinero transferido a " + persona.nombre + ": " +  str(dinero_transferir) + ". Saldo: " + str(self.saldo)
        else:
            return "Saldo insuficiente. " + "Saldo: " + str(self.saldo)
    def agregar_dinero(self, dinero_agregar):
        self.saldo += dinero_agregar
        return "Saldo: " + str(self.saldo)
        
user1 = UsuarioBanco("Alicia", 100, True)
user2 = UsuarioBanco("Bob", 50, True)

print(user2.agregar_dinero(20))
print(user2.transferir_dinero(80, user1))
print(user1.retirar_dinero(50))
```

## Ejercicio 36

```python
def contar_palabras(texto):
    dicts_cont_pal = {}
    list_pal = []
    palabra = ""

    for letra in texto:
        if letra == " ":
            list_pal.append(palabra)
            palabra = ""
        else:
            palabra += letra

    list_pal.append(palabra)
    
    for palabra in list_pal:
        if palabra not in dicts_cont_pal:
            dicts_cont_pal[palabra] = 1
        else:
            dicts_cont_pal[palabra] += 1

    return dicts_cont_pal

def reemplazar_palabras(palabra_original, palabra_nueva, texto):
    return texto.replace(palabra_original, palabra_nueva)

def eliminar_palabra(palabra_eliminar, texto):
    return texto.replace(palabra_eliminar, "")


def procesar_texto(texto, opcion, *args):
    if opcion.lower() == "contar":
        return contar_palabras(texto)
    elif opcion.lower() == "reemplazar":
        return reemplazar_palabras(args[0], args[1], texto)
    elif opcion.lower() == "eliminar":
        return eliminar_palabra(args[0], texto)
    else:
        return "Selecciona una opción válida"
    
texto = "Hola Hola soy Jairo Martín"

print(procesar_texto(texto, "contar"))
print(procesar_texto(texto, "reemplazar", "Jairo", "Pablo"))
print(procesar_texto(texto, "eliminar", "Martín"))
```

## Ejercicio 39

```python
import math 
def calcular_area(figura, *args):
    if figura.lower() == "rectangulo":
        return args[0] * args[1]
    elif figura.lower() == "triangulo":
        return (args[0] * args[1]) / 2
    elif figura.lower() == "circulo":
        return 2 * math.pi * (args[0] * args[0])
    else:
        return "Seleccione una figura válida"
    
print(calcular_area("circulo", 3))
print(calcular_area("triangulo", 2, 5))
print(calcular_area("rectangulo", 3, 4))
```
